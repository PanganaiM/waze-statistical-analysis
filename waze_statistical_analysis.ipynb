# Waze User Behaviour — Statistical Analysis & Hypothesis Testing

A portfolio project examining whether mean driving activity differs between iPhone and Android Waze users.

## 1. Project Overview

This analysis uses descriptive statistics and an independent two-sample hypothesis test to investigate whether device type is associated with a difference in the mean number of drives. The analysis is intended to support evidence-based interpretation rather than relying on the small observed difference in group averages.

## 2. Business Question

**Is there a statistically significant difference in the mean number of drives between iPhone and Android users in the Waze dataset?**

## 3. Tools & Dataset

- Python
- pandas
- SciPy
- Jupyter Notebook
- `waze_dataset.csv`

## 4. Data Loading

```python
# Import any relevant packages or libraries
import pandas as pd
from scipy import stats
```

```python
# Load dataset into dataframe
df = pd.read_csv('waze_dataset.csv')
```

## 5. Data Preparation

Device type is retained as the original categorical variable and also mapped to a numeric representation for analysis without overwriting the source field.

```python
# 1. Create `map_dictionary`
map_dictionary = {'Android': 2, 'iPhone': 1}

# 2. Create new `device_type` column
df['device_type'] = df['device'].copy()

# 3. Map the new column to the dictionary
df['device_type'] = df['device_type'].map(map_dictionary)
df[['device', 'device_type']].head()
```

## 6. Descriptive Statistics

The first comparison examines average drives by device type.

```python
df.groupby('device')['drives'].mean()
```

The observed means are approximately **67.86 drives for iPhone users** and **66.23 drives for Android users**, a difference of about **1.63 drives**. A hypothesis test is required to determine whether this difference is statistically significant rather than a result of sampling variation.

## 7. Hypotheses and Significance Level

- **Null hypothesis (H₀):** The mean number of drives is equal for iPhone and Android users.
- **Alternative hypothesis (H₁):** The mean number of drives differs between iPhone and Android users.
- **Significance level:** α = 0.05

Because the two device groups are independent and equal population variances are not assumed, Welch's independent two-sample t-test is used.

## 8. Welch's Two-Sample t-Test

```python
# 1. Isolate the `drives` column for iPhone users.
iphone_drives = df.loc[df['device'] == 'iPhone', 'drives']

# 2. Isolate the `drives` column for Android users.
android_drives = df.loc[df['device'] == 'Android', 'drives']

# 3. Perform the t-test
alpha = 0.05
t_stat, p_value = stats.ttest_ind(iphone_drives, android_drives, equal_var=False)
print('t-statistic:', t_stat)
print('p-value:', p_value)
```

## 9. Statistical Interpretation

The test returns **t ≈ 1.464** and **p ≈ 0.1434**. Since the p-value is greater than 0.05, the analysis **fails to reject the null hypothesis**. There is insufficient statistical evidence in this dataset to conclude that mean drives differ between iPhone and Android users.

## 10. Business Interpretation

Device type alone does not provide evidence of a meaningful difference in average driving activity. Although iPhone users show a slightly higher sample mean, the difference is not statistically significant at the 5% level. Decisions about engagement or retention should therefore consider stronger behavioural variables rather than assuming that device platform explains driving frequency.

## 11. Limitations & Next Steps

This test evaluates one relationship: device type versus mean drives. It does not establish causation and does not account for other user characteristics. Further work can examine churn, activity days, driving days, tenure, distance and multivariable predictive models.

## 12. Conclusion

The statistical analysis demonstrates how descriptive statistics and hypothesis testing can distinguish an observed sample difference from a statistically supported difference. For this dataset, the evidence does not support a significant difference in mean drives between iPhone and Android users.
